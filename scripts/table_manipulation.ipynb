```python
import pandas as pd
```

```python
# розведення загальної таблиці на окремі аркуші (відповідно до людей-анотаторів)

input_path = r"..."
output_path = r"..."

df = pd.read_excel(input_path)

with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
    for name, group in df.groupby("from"):
        sheet_name = str(name).strip()[:31]

        group.to_excel(writer, sheet_name=sheet_name, index=False)
```
